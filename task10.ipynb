```python
def pos_tag_and_extract_info(text):
  doc = nlp(text)
  nouns = []
  verbs = []
  adjectives = []
  entities = [(ent.text, ent.label_) for ent in doc.ents]

  for token in doc:
    if token.pos_ == "NOUN":
      nouns.append(token.text)
    elif token.pos_ == "VERB":
      verbs.append(token.text)
    elif token.pos_ == "ADJ":
      adjectives.append(token.text)

  return {
    "nouns": nouns,
    "verbs": verbs,
    "adjectives": adjectives,
    "entities": entities
  }
```

```python
# The correct function is defined in cell FBGZurQFlzsQ.
# Let's test it here with some example text:

sample_text = "Google Colab is a hosted Jupyter notebook service that requires no setup to use."
results = pos_tag_and_extract_info(sample_text)

print("Nouns:", results["nouns"])
print("Verbs:", results["verbs"])
print("Adjectives:", results["adjectives"])
print("Entities:", results["entities"])
```

```python
web_document = """
<html>
<head>
<title> Example Web Page</title>
</head>
<body>
<p>This is an example web page. It contains some text with various parts of speech.</p>
<p>For example, "The cat jumps over the lazy dog" contains a noun, a verb, and prepositions. </p>
</body>
</html>
"""
```

```python
def extract_text_from_html(html):
  soup = BeautifulSoup(html, 'html.parser')
  return soup.get_text()
```

```python
text_content = extract_text_from_html(web_document)
results = pos_tag_and_extract_info(text_content)

print("Nouns:", results["nouns"])
print("Verbs:", results["verbs"])
print("Adjectives:", results["adjectives"])
print("Entities:", results["entities"])
```
